```python
# Load and Inspect: OTT Viewer Engagement Analysis

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the data
df = pd.read_csv("../data/raw/ott_data.csv")
print("First 5 rows:")
display(df.head())

# 2. Size and column names
print("\nShape (rows, columns):", df.shape)
print("Columns:", df.columns.tolist())

# 3. Data types and non-null counts
print("\nInfo:")
df.info()

# 4. Summary statistics
print("\nSummary statistics:")
display(df.describe())

# 5. Missing values
print("\nMissing values per column:")
print(df.isnull().sum())

# 6. Duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())
display(df[df.duplicated(keep=False)].sort_values("Viewer_ID"))

# 7. Unique values of category columns
for col in ["City", "Subscription_Plan", "Device", "Preferred_Genre"]:
    print(f"\n{col}:")
    print(df[col].value_counts(dropna=False))
```
